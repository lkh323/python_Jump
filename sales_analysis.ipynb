```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 그래프 스타일 설정 (깔끔한 디자인)
plt.style.use('seaborn-v0_8-whitegrid')

print("라이브러리 로드 완료!")
```

```python
# 랜덤 시드 고정 (항상 같은 결과가 나오도록 설정)
np.random.seed(42)

# 데이터 생성
months = ['1월', '2월', '3월', '4월', '5월', '6월', 
          '7월', '8월', '9월', '10월', '11월', '12월']

data = {
    '월': months,
    '매출': np.random.randint(1000, 5000, 12),  # 1000~5000 사이 랜덤 값
    '비용': np.random.randint(500, 2000, 12)   # 500~2000 사이 랜덤 값
}

# 데이터프레임(표) 만들기
df = pd.DataFrame(data)

# 순이익 계산 (매출 - 비용)
df['순이익'] = df['매출'] - df['비용']

# 상위 5개 데이터 확인 (즉시 결과 출력 체험)
df.head()
```

```python
# 데이터 통계 요약
print("=== 데이터 통계 요약 ===")
df.describe()
```

```python
# 한글 폰트 설정이 안 되어 있을 경우를 대비해 영문으로 라벨링합니다.
plt.figure(figsize=(12, 6))

# 매출 꺾은선 그래프
plt.plot(df['월'], df['매출'], marker='o', linestyle='-', linewidth=2, label='Sales (Revenue)')

# 순이익 막대 그래프
plt.bar(df['월'], df['순이익'], alpha=0.5, color='green', label='Net Profit')

plt.title('Monthly Sales & Profit Analysis', fontsize=15)
plt.xlabel('Month')
plt.ylabel('Amount ($)')
plt.legend()
plt.grid(True, alpha=0.3)

# 그래프 출력
plt.show()
```
